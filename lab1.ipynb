## Задача 1.8.

 > В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы? 

* а) Рассмотрим событие $A$ - по каждой номинации установлены различные призы

$P(A) = A_{10}^{5} = 10^{5}$



```python
round(10 ** 5)
```

* б) Рассмотрим событие $B$ - по каждой номинации установлены одинаковые призы

$P(B) = C_{10}^5 = C_{10+5-1}^6 = C_{14}^5$

```python
round(sc.comb(14,5))
```

## Задача 1.9.
> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза? 


$P_7(3; 2; 2) = \frac{7!}{3! \cdot 2! \cdot 2!}$

```python
sc.factorial(7)/(sc.factorial(3)*sc.factorial(2)*sc.factorial(2))
```

## Задача 1.10.
> Буквы Т, Е, И, Я, Р, О написаны на отдельных
карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

* а) Рассмотрим событие $A$ - получение слова «ТОР»

$P(A) = \frac{m}{n} = \frac{1}{A_6^3} = \frac{1}{6 \cdot 5 \cdot 4}$

```python
round((1/(6*5*4)),4)
```

* б) Рассмотрим событие $B$ - получение слова «ТЕОРИЯ»

$P(B) = \frac{m}{n} = \frac{1}{P_6} = \frac{1}{6!}$

```python
round((1/sc.factorial(6)),4)
```

## Задача 1.11.
> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных
карточках написаны три буквы А, две буквы Н и одна буква С.

* Рассмотрим событие $B$ - получение слова «АНАНАС»

$P(B) = 1 : P_6(3; 2; 1) = 1 : \frac{6!}{3! \cdot 2! \cdot 1!}$

```python
round((1/(sc.factorial(6)/(
    sc.factorial(3)
    * sc.factorial(2)
    * sc.factorial(1)))),4
)
```

## Задача 1.16.
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар. 

* а) Рассмотрим событие $А$ - никакие два лица одного
пола не сядут рядом.

$P(A) = \frac{m_1}{n} = \frac{2 \cdot (5!)^2}{10!}$

```python
round((2*(sc.factorial(5)**2)/sc.factorial(10)),4)
```

* б) Рассмотрим событие В - супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар:

$P(B) = \frac{m_2}{n} = \frac{5! \cdot 2^5}{10!}$

```python
round((sc.factorial(5)*(2**5)/sc.factorial(10)),4)
```

## Задача 1.17.
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

* а) Рассмотрим событие $А$ - пассажиры попали в два купе.

$P(A) = \frac{m_1}{n} = \frac{C_9^2 \cdot C_8^7}{C_{36}^7}$

```python
f'{(round((sc.comb(9,2)*sc.comb(8,7)/sc.comb(36,7)),6)):.6f}'
```

* б) Рассмотрим событие $B$ - пассажиры попали в семь купе

$P(B) = \frac{m_2}{n} = \frac{C_{9}^7 \cdot 4^7}{C_{36}^7}$

```python
round((sc.comb(9,7)*(4**7)/sc.comb(36,7)),4)
```

* в) Рассмотрим событие $C$ - пассажиры попали в три купе.

$P(C) = \frac{m_3}{n} = \frac{C_{9}^3(6C_4^4C_4^2C_4^1 + 3C_4^3C_4^3C_4^1 + 3C_4^3C_4^2C_4^2)}{C_{36}^7}$



```python
round((sc.comb(9,3)*(6*sc.comb(4,4)*sc.comb(4,2)*sc.comb(4,1)
                     + 3*sc.comb(4,3)*sc.comb(4,3)*sc.comb(4,1)
                     + 3*sc.comb(4,3)*sc.comb(4,2)*sc.comb(4,2))
       /sc.comb(36,7)),5)
```
